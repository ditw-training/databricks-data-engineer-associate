## Must-Know Syntax

Essential SQL and Python syntax patterns that appear frequently on the certification exam.

### MERGE INTO
Upsert pattern to insert or update rows based on a matching condition.
```sql
MERGE INTO target USING source ON target.id = source.id
WHEN MATCHED THEN UPDATE SET *
WHEN NOT MATCHED THEN INSERT *;
```

### Auto Loader
Incrementally ingest new files from cloud storage using Structured Streaming.
```python
spark.readStream.format("cloudFiles")
    .option("cloudFiles.format", "json")
    .option("cloudFiles.schemaLocation", checkpoint)
    .load(path)
```

### Lakeflow Expectations
Data quality constraints applied to streaming tables in Lakeflow Spark Declarative Pipelines.
```sql
CONSTRAINT valid_id EXPECT (id IS NOT NULL) ON VIOLATION DROP ROW
```

### Change Data Feed
Track row-level changes (inserts, updates, deletes) in a Delta table.
```sql
ALTER TABLE t SET TBLPROPERTIES ('delta.enableChangeDataFeed' = true);
SELECT * FROM table_changes('t', 2);
```

### GRANT
Assign privileges on Unity Catalog objects to users or groups.
```sql
GRANT USE CATALOG ON CATALOG my_catalog TO `analysts`;
GRANT USE SCHEMA ON SCHEMA my_catalog.silver TO `analysts`;
GRANT SELECT ON SCHEMA my_catalog.silver TO `analysts`;
```

### Declarative Automation Bundles — CLI
Validate, deploy and run a bundle target from the folder containing `databricks.yml`.
```bash
databricks bundle validate -t dev --var="catalog=my_catalog"
databricks bundle deploy   -t dev --var="catalog=my_catalog"
databricks bundle run      -t dev --var="catalog=my_catalog" my_job   # resource key from the YAML
```

### Column Mask & Row Filter
Attach a masking UDF to a column and a filtering (BOOLEAN) UDF to a table.
```sql
ALTER TABLE t ALTER COLUMN email SET MASK mask_email;
ALTER TABLE t SET ROW FILTER region_filter ON (region);
```

### AUTO CDC (Lakeflow SQL)
Apply change events to a target streaming table, keeping full history (SCD Type 2).
```sql
CREATE OR REFRESH STREAMING TABLE customers_scd2;
CREATE FLOW customers_cdc AS AUTO CDC INTO customers_scd2
FROM STREAM customers_raw
KEYS (customer_id)
SEQUENCE BY updated_at
STORED AS SCD TYPE 2;
```

### Lakeflow Jobs — `for_each` & task values
Loop a task over inputs (`{{input}}` = current element) and pass small values between tasks.
```yaml
- task_key: process_tables
  for_each_task:
    inputs: '["orders", "customers"]'
    task:
      task_key: process_one
      notebook_task:
        notebook_path: ../src/process_table
        base_parameters:
          table: "{{input}}"
```
```python
dbutils.jobs.taskValues.set(key="row_count", value=1000)                          # upstream task
n = dbutils.jobs.taskValues.get(taskKey="validate", key="row_count", default=0)   # downstream task
# in task parameters: {{tasks.validate.values.row_count}}
```

## Common Exam Traps

Frequently tested gotchas and misconceptions — review these carefully before the exam.

| Trap | Correct Answer |
|------|---------------|
| VACUUM default retention | **7 days** (168 hours) |
| Streaming vs Materialized View | `CREATE STREAMING TABLE` = incremental — each input row processed once (append-only source or AUTO CDC target); `CREATE MATERIALIZED VIEW` = result always reflects its query — refreshed incrementally when possible, otherwise fully recomputed |
| Expectation without `ON VIOLATION` | **Warn only** — keeps all rows |
| `DROP ROW` vs `FAIL UPDATE` | DROP ROW silently removes; FAIL UPDATE aborts |
| Temp View scope | Current SparkSession only |
| Global Temp View | Query as `global_temp.view_name` |
| Auto Loader format | `cloudFiles` (not `autoLoader`!) |
| Unity Catalog default | **Deny by default** — must GRANT explicitly |
| OPTIMIZE vs VACUUM | OPTIMIZE compacts files; VACUUM removes old files |
| Schema evolution | `.option("mergeSchema", "true")` on **write**, not read |
| Change Data Feed | `table_changes()`, NOT `table_change_feed()` |
| Streaming trigger | `availableNow=True` = all then stop; `processingTime` = micro-batch |
| Row Filter function | Must return **BOOLEAN** |
| Column Mask function | Must return **same type** as masked column |
