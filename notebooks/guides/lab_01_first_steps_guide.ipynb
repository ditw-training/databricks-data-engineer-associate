# Guide: LAB 01 — Platform & Workspace First Steps

**Duration:** ~25 min | **Difficulty:** ⭐ Beginner | **Lab notebook:** [lab_01_first_steps](../day1/lab/lab_01_first_steps.ipynb)

## Scenario

It's your first day on the RetailHub data team. Before building anything, you get comfortable with the workspace: attach the notebook to **Serverless** compute, explore your **Unity Catalog** environment, create a **Volume**, and run your first Spark reads.

## Key Concepts

| Concept | What to remember |
|---|---|
| **Serverless compute** | Default for notebooks/jobs/SQL — instant start, Databricks-managed. Classic all-purpose clusters are shown only as an instructor demo. |
| **Unity Catalog namespace** | 3 levels: `catalog.schema.object`. Your catalog is `retailhub_<your_slug>`. |
| **Volumes** | Governed file storage. **Managed** = metastore-managed location; **external** = existing cloud path. Path pattern: `/Volumes/<catalog>/<schema>/<volume>/...` |
| **`%run` setup** | Every lab starts with `%run ../../setup/00_setup`, which exports `CATALOG`, `BRONZE_SCHEMA`, `SILVER_SCHEMA`, `GOLD_SCHEMA`, `DATASET_PATH`. |

## Prerequisites

- Trainer ran `setup/00_pre_config` (your catalog and datasets exist).
- You can attach the notebook to **Serverless** (Connect → Serverless).
- No cluster creation needed — that part is an instructor demo only.

## Tasks Overview

| # | Task | Hint area |
|---|---|---|
| 1 | Verify your catalog & schemas | `current_catalog()`, `SHOW SCHEMAS` |
| 2 | Create a managed Volume with SQL | `CREATE VOLUME IF NOT EXISTS ...` |
| 3 | Write & read a file with `dbutils.fs` | `dbutils.fs.put` / `dbutils.fs.head` |
| 4 | Read a dataset CSV with Spark | `spark.read` with options |

## Task 1 Hints — Verify Your Catalog & Schemas

- `spark.sql("SELECT current_catalog()")` returns a 1-row DataFrame — use `.first()[0]`.
- The setup notebook already ran `USE CATALOG` for you; compare the result with the exported `CATALOG` variable.
- List schemas with `SHOW SCHEMAS IN <catalog>` and check that `bronze`, `silver`, `gold` are present (collect the column into a Python set).

## Task 2 Hints — Create a Managed Volume

- Syntax: `CREATE VOLUME IF NOT EXISTS <catalog>.<schema>.<name>` — no `LOCATION` clause means **managed**.
- Use an f-string with `CATALOG` and `BRONZE_SCHEMA` so the code works for every participant.
- Verify via `information_schema.volumes` (filter by `volume_name`) — the validation also checks `volume_type = 'MANAGED'`.

## Task 3 Hints — Write and Read a File

- Volume paths look like `/Volumes/<catalog>/<schema>/<volume>/file.txt`.
- `dbutils.fs.put(path, content, overwrite=True)` writes a small text file; `dbutils.fs.head(path)` reads it back.
- The validation compares the exact string you wrote — don't add extra whitespace.

## Task 4 Hints — Read a Dataset File with Spark

- The training datasets live under `DATASET_PATH` (a Volume) — e.g. `f"{DATASET_PATH}/customers/customers.csv"`.
- CSV needs `header=True`; `inferSchema` is fine for a first look (you'll do explicit schemas in LAB 02).
- `df.count()` must be greater than 0 — 10,000 customers are expected.

## Common Pitfalls

- ❌ Skipping `%run ../../setup/00_setup` — the `CATALOG`/`DATASET_PATH` variables won't exist.
- ❌ Creating the volume in `default` instead of the **bronze** schema.
- ❌ Using a 2-level name (`schema.volume`) — Unity Catalog needs 3 levels unless `USE CATALOG`/`USE SCHEMA` are set.
- ❌ Trying to create a classic cluster — this training is Serverless-first; the cluster UI is instructor-demo only.
