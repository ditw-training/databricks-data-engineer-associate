## RetailHub Scenario

Throughout the training we build a **RetailHub** data platform — an e-commerce company processing orders, customers, and products.

You will:
- Ingest raw CSV/JSON data into the Bronze layer
- Transform and cleanse data into Silver
- Build aggregated Gold tables for analytics
- Automate the pipeline with Lakeflow and Jobs
- Secure it with Unity Catalog governance
