## Answer Key -- Day 3

| # | Ans | Explanation |
|---|-----|-------------|
| Q1 | **C** | Bronze is the raw/landing layer. Silver is cleaned/validated. Gold is business-level aggregations. |
| Q2 | **B** | A STREAMING TABLE processes each input row once, incrementally (append-only sources). A MATERIALIZED VIEW stores the result of its query and keeps it correct on each refresh (incrementally where possible, otherwise by full recomputation). |
| Q3 | **D** | In current pipelines, datasets in the same pipeline are referenced by name (or fully qualified name) — no prefix is needed. `LIVE.` is legacy syntax, not required. |
| Q4 | **C** | `ON VIOLATION DROP ROW` removes rows that fail the expectation from the target; the number of dropped rows is recorded in the pipeline's data quality metrics. |
| Q5 | **D** | `FAIL UPDATE` aborts the pipeline immediately, preventing bad data from being written. |
| Q6 | **C** | Without `ON VIOLATION`, the expectation is in "warn" mode: all rows pass through, but violations are logged as metrics. |
| Q7 | **B** | `STREAM(table_name)` reads from an upstream dataset as a streaming source. Without `STREAM`, the dataset is read as a batch (as in a MATERIALIZED VIEW). |
| Q8 | **D** | `event_log(TABLE(...))` is the function to query pipeline event logs, including data quality metrics. |
| Q9 | **B** | A Task is the smallest unit of work in a Lakeflow Job. Jobs can have multiple tasks with dependencies forming a DAG. |
| Q10 | **A** | `taskValues.set(key, value)` stores a value that downstream tasks retrieve with `taskValues.get()`. |
| Q11 | **B** | Repair Run re-executes only failed tasks and their downstream dependents, saving time and compute. |
| Q12 | **A** | Valid task types include: Notebook, SQL, Python script, Lakeflow pipeline, JAR, dbt, and more. |
| Q13 | **C** | A row filter is a SQL UDF attached to the table (`ALTER TABLE ... SET ROW FILTER`), so every query is filtered by the user's identity or group. A dynamic view filters only queries that go through the view. |
| Q14 | **D** | Column Masks are applied via `ALTER TABLE ... ALTER COLUMN ... SET MASK <function>`. |
| Q15 | **B** | `INFORMATION_SCHEMA.TABLES` lists all tables and views with their schema, name, and type. |
| Q16 | **A** | `is_account_group_member('group_name')` returns true if the current user belongs to the specified group. |
| Q17 | **C** | Deliveries are irregular, so a data-driven trigger fits: file arrival starts a run when new files land in the volume and does nothing otherwise. Time-based (scheduled) triggers suit fixed deadlines regardless of data; table update triggers suit upstream tables; continuous keeps a run always active. |
| Q18 | **A** | `GRANT SELECT ON SCHEMA catalog.schema TO principal` grants SELECT on all current and future tables in the schema (the group also needs `USE CATALOG` and `USE SCHEMA`). `GRANT ALL PRIVILEGES` would grant far more than read access. |
| Q19 | **B** | `ALTER TABLE t DROP ROW FILTER` removes the row filter. Similarly, `ALTER TABLE t ALTER COLUMN c DROP MASK` removes a mask. |
| Q20 | **A** | Git folders let you create and switch branches and **Commit & Push** from the workspace UI. Pull requests are created, reviewed and merged in the Git provider (GitHub, Azure DevOps, GitLab). Bundles deploy code; they do not merge branches. |
| Q21 | **D** | Errors that start right after a library change on a healthy, running cluster point to a library conflict. Remove the conflicting cluster library and use notebook-scoped `%pip install` with pinned versions (or serverless environments). Startup failures are diagnosed in the compute event log; OOM shows `OutOfMemoryError` or spill. |
| Q22 | **B** | `ALTER TABLE ... SET MANAGED` converts an external Delta table to a managed table in place, keeping name, history and permissions (Databricks Runtime 17.3 LTS+ or serverless, table owner). `UNSET MANAGED` rolls back within 14 days. A clone creates a new table; dropping an external table leaves its files and D recreates it as external. |
| Q23 | **C** | The job's run history shows durations of past runs, and the matrix view shows per-task durations across runs, so you can compare today against the usual baseline and find the task that slowed down (`system.lakeflow.job_run_timeline` gives the same history in SQL). |
| Q24 | **A** | Managed connectors (Salesforce, Workday, SQL Server...) are exactly for SaaS/DB sources with managed CDC — minimal code, UC-governed targets. Auto Loader/COPY INTO are for files already in object storage. |
| Q25 | **D** | Bundles define targets (dev/test/prod) with variable overrides so one codebase deploys everywhere via `databricks bundle deploy -t <target>`. |
| Q26 | **C** | A few long-running tasks with outsized shuffle read is the classic data-skew signature at the stage level. |
| Q27 | **A** | ABAC applies row filters/column masks centrally through policies bound to governed tags — tag once, policy applies everywhere. |
| Q28 | **D** | Predictive optimization (default-on for UC managed tables) plus Liquid Clustering / `CLUSTER BY AUTO` replaces manually scheduled OPTIMIZE/ZORDER/VACUUM as the recommended answer. |
