## Q20

Which statement correctly reads a JSON file with an explicit schema in PySpark?

**A.**
```python
df = spark.read.json("/path/data.json").schema("id INT, name STRING")
```

**B.**
```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType
schema = StructType([StructField("id", IntegerType()), StructField("name", StringType())])
df = spark.read.schema(schema).json("/path/data.json")
```

**C.**
```python
df = spark.read.format("json").option("schema", "id:int, name:string").load("/path")
```

**D.**
```python
df = spark.read.json("/path").cast("id INT, name STRING")
```

## Answer Key -- Day 1

| # | Ans | Explanation |
|---|-----|-------------|
| Q1 | **C** | Unity Catalog provides centralized access control, auditing, lineage, and data discovery across workspaces. |
| Q2 | **A** | The hierarchy is: Metastore > Catalog > Schema > Table/View/Function. |
| Q3 | **C** | Reading a table needs `USE CATALOG` on the catalog, `USE SCHEMA` on the schema and `SELECT` on the table (or inherited from a parent). `USE CATALOG`/`USE SCHEMA` alone only allow access to the containers; `BROWSE` shows metadata only. |
| Q4 | **D** | `CREATE TABLE ... AS SELECT * FROM read_files()` (CTAS with `read_files`) is the standard pattern for loading files into managed Delta tables. |
| Q5 | **C** | Managed tables use catalog-managed storage; dropping them deletes the underlying data. External tables reference external locations and survive DROP. |
| Q6 | **A** | `createOrReplaceTempView()` registers a DataFrame as a session-scoped temporary view queryable via `spark.sql()`. |
| Q7 | **B** | The Delta transaction log (`_delta_log`) provides ACID guarantees, enables time travel, and enforces schema on write. |
| Q8 | **A** | Delta Lake stores data as Parquet files with an additional JSON-based transaction log. |
| Q9 | **B** | Each version in Delta History corresponds to a single transactional operation on the table. |
| Q10 | **A** | `VERSION AS OF` enables time travel, returning the table snapshot at the specified version number. |
| Q11 | **D** | Serverless SQL warehouses start in seconds, scale out for concurrent BI queries, stop when idle and need no infrastructure management. All-purpose clusters are for interactive notebooks; job clusters are for automated jobs. |
| Q12 | **C** | Unity Catalog Volumes are the recommended replacement for DBFS root/mounts for storing non-tabular files. |
| Q13 | **D** | ELT (Extract-Load-Transform) loads raw data into the lakehouse first, then transforms in-place using Spark. ETL transforms before loading. Databricks favors ELT. |
| Q14 | **B** | `DESCRIBE DETAIL` returns physical metadata: location, size, numFiles, partitioning/clustering columns, properties, createdAt, etc. `DESCRIBE TABLE` lists columns and types. |
| Q15 | **D** | `VACUUM RETAIN 0 HOURS` (only possible after disabling the safety check) removes every data file not referenced by the current version, so time travel to older versions fails. |
| Q16 | **B** | `%run` executes another notebook in the same context, making its variables and functions available to the calling notebook. |
| Q17 | **D** | By default, Delta enforces the schema -- the write fails. Adding `.option("mergeSchema", "true")` enables schema evolution and adds the column. |
| Q18 | **A** | `dbutils.fs` provides file system utilities: `ls()`, `cp()`, `rm()`, `head()`, `mkdirs()`. |
| Q19 | **C** | `RESTORE TABLE` reverts the table to a previous state and creates a new version entry in the history log. |
| Q20 | **B** | Define a `StructType` schema and pass it via `.schema(schema)` before calling `.json()` (a DDL string such as `.schema("id INT, name STRING")` on the reader also works). A fails: the schema must be set on the reader, and `DataFrame.schema` is a property, not a method. C and D are not valid APIs. |
