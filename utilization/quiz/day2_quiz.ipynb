## Answer Key -- Day 2

| # | Ans | Explanation |
|---|-----|-------------|
| Q1 | **A** | `OPTIMIZE` compacts small files (small file problem) into larger files, improving read performance. |
| Q2 | **C** | Tables smaller than `spark.sql.autoBroadcastJoinThreshold` (10 MB by default) are broadcast automatically. Raising it above the dimension size switches to a broadcast hash join; always re-measure (plan shows `BroadcastHashJoin`, run time drops). `-1` disables auto-broadcast. |
| Q3 | **B** | `VACUUM` removes stale data files no longer part of the current table version. Default retention is 7 days. |
| Q4 | **C** | Liquid Clustering (`CLUSTER BY`) automatically optimizes data layout, replacing manual partitioning and ZORDER. |
| Q5 | **A** | Directory listing (the default) lists the input path on every run, which gets slow with millions of files. File notification mode discovers new files from storage events; Databricks recommends file events (`cloudFiles.useManagedFileEvents`) over the legacy `cloudFiles.useNotifications`. |
| Q6 | **B** | Temporary views exist only within the SparkSession (lost when the session ends). Permanent views are registered in Unity Catalog and can be queried by anyone with privileges. |
| Q7 | **C** | The checkpoint directory stores offset info and state, ensuring exactly-once fault-tolerant processing. |
| Q8 | **B** | `cloudFiles.format` specifies the source file format (e.g., `json`, `csv`, `parquet`) for Auto Loader. |
| Q9 | **D** | `availableNow=True` processes all currently available data in incremental batches and then stops. Ideal for scheduled jobs. |
| Q10 | **C** | `COPY INTO` is idempotent SQL that records loaded files in the target table's Delta log. Auto Loader keeps file state in its checkpoint and scales to millions of files with directory listing or file notification mode. |
| Q11 | **A** | Auto Loader uses `format("cloudFiles")` with `readStream`, and requires `cloudFiles.format` + `cloudFiles.schemaLocation` options. |
| Q12 | **D** | Predictive Optimization automatically schedules and runs OPTIMIZE, VACUUM, and ANALYZE based on usage patterns. |
| Q13 | **B** | Deletion Vectors mark rows as deleted in a side file without rewriting the entire Parquet file, speeding up DELETE/UPDATE/MERGE. |
| Q14 | **D** | A window function with `ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW` creates a running total partitioned by category. |
| Q15 | **A** | `explode()` takes an array or map column and generates one row per element (or key-value pair). |
| Q16 | **B** | `UNION ALL` appends all rows and keeps duplicates. `UNION` (= `UNION DISTINCT`) removes duplicates at the cost of an extra shuffle. `CROSS JOIN` is a Cartesian product. In PySpark, `union()`/`unionAll()` keep duplicates and match columns by position; `unionByName()` matches by name. |
| Q17 | **D** | `ROW_NUMBER()` gives each row a unique sequential number (1, 2, 3...) with no gaps, regardless of ties. |
| Q18 | **C** | `approx_count_distinct` (HyperLogLog++) is a fast single-pass estimate for very large data. `summary()` returns count, mean, stddev, min, 25%, 50%, 75% and max; `describe()` omits the quartiles. |
| Q19 | **A** | One straggler task with shuffle read far above the median is the classic skew signature. AQE skew join handling (`spark.sql.adaptive.skewJoin.enabled`, on by default) splits the skewed partition in sort-merge joins; salting is the manual alternative. |
| Q20 | **D** | `dropDuplicates(["col1", "col2"])` removes duplicate rows based on a subset of columns. `distinct()` works on all columns. |
